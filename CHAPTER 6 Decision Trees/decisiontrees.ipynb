```python
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier
```

```python
iris = load_iris()
X= iris.data[:, 2:] # petal length and width
y = iris.target
```

```python
tree_clf = DecisionTreeClassifier(max_depth=2)
tree_clf.fit(X, y)
```

```python
from sklearn.tree import export_graphviz

export_graphviz(
    tree_clf,
    out_file="iris_tree.dot",
    feature_names=iris.feature_names[2:],
    class_names=iris.target_names,
    rounded=True,
    filled=True
)
```

```python
import graphviz
from IPython.display import Image

# Display the saved PNG
Image("iris_tree.png")
```

```python
tree_clf.predict_proba([[5, 1.5]])
```

```python
tree_clf.predict([[5, 1.5]])
```

```python
from sklearn.tree import DecisionTreeRegressor
tree_reg = DecisionTreeRegressor(max_depth=2)
tree_reg.fit(X, y)
```

```python
from sklearn.tree import export_graphviz

export_graphviz(
    tree_reg,
    out_file="iris_tree_reg.dot",
    feature_names=iris.feature_names[2:],
    rounded=True,
    filled=True
)

```

```python
import graphviz
from IPython.display import Image

# Read the .dot file and render to png
graph = graphviz.Source.from_file("iris_tree_reg.dot")
graph.render("iris_tree_reg", format="png", cleanup=True)  # Creates iris_tree_reg.png

# Display the saved PNG
Image("iris_tree_reg.png")
```
